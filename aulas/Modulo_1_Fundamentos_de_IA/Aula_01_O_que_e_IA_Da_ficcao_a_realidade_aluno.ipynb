**Exercício da Aula 01**  
### Mapeando a IA no seu dia a dia ★ ★ ★

*Em dupla · 15 minutos · Sem computador · Entrega: papel ou caderno*

1. Liste 5 ferramentas digitais que você usou nas últimas 24 horas. Qualquer categoria conta: mensageiro, rede social, mapas, streaming de vídeo ou música, banco, delivery, teclado do celular, buscador.
2. Classifique cada uma no mapa visto no slide 6: é IA sem aprendizado (regra/busca), ML clássico, Deep Learning ou IA Generativa? Escreva a camada ao lado de cada ferramenta. Se a mesma ferramenta usar duas camadas, anote as duas.
3. Escolha a mais interessante e descreva o mecanismo em 2 frases — o que entra, o que sai, e o que o sistema precisa ter aprendido para funcionar. Não precisa estar certo; precisa ser específico.
4. Aponte 1 risco concreto: o que acontece se essa IA errar? Descreva a falha (o quê ), quem é afetado (quem ) e se a pessoa afetada teria como perceber o erro (visibilidade ). Esse terceiro item costuma ser o mais desconfortável.
5. Formule 1 pergunta que você gostaria de fazer ao time que construiu essa ferramenta. Guarde a pergunta: vamos retomá-la na Aula 03.

> **💡 Dica:**
>
> não existe resposta errada aqui — o exercício é de raciocínio, não de conhecimento técnico. Sua hipótese sobre o funcionamento pode estar completamente errada, e isso é exatamente o que torna a discussão útil. O único jeito de fazer errado é responder de forma genérica ("usa IA", "pode dar problema").

## 📚 Referências da aula

- Livro Russell, S.; Norvig, P. — Inteligência Artificial: Uma Abordagem Moderna . 3ª ed. Pearson, 2016. Cap. 1 — definição do campo, histórico e as quatro escolas de pensamento em IA.
- Livro Alpaydin, E. — Aprendizado de Máquina . Pearson, 2016. Cap. 1 — por que aprender com dados difere de programar regras.
- Livro Goodfellow, I.; Bengio, Y.; Courville, A. — Deep Learning . Pearson, 2017. Cap. 1 — o salto do ML clássico para redes profundas e por que ele dependeu de dados e GPU.
- Livro Chollet, F. — Deep Learning com Python . Pearson, 2018. Cap. 1 — leitura de apoio para quem quiser a intuição visual das redes neurais.
- Paper Vaswani, A. et al. — "Attention Is All You Need". NeurIPS, 2017. O paper que introduziu o Transformer. arxiv.org/abs/1706.03762
- Paper Brown, T. et al. — "Language Models are Few-Shot Learners". NeurIPS, 2020. Mostrou que escala + exemplos no prompt substituem retreinamento. Base das Aulas 05 e 06. arxiv.org/abs/2005.14165
- Paper Turing, A. — "Computing Machinery and Intelligence". Mind , 1950. O artigo que fundou o campo com a pergunta "pode uma máquina pensar?".
- Online DAIR.AI — Prompt Engineering Guide . Referência principal da disciplina, seção "Introduction". promptingguide.ai
- Online Anthropic — Engineering Blog e Learn : fundamentos de prompting e context engineering. anthropic.com/engineering
- Online OpenAI — Developer documentation e cookbook : referência de API e boas práticas de prompt. developers.openai.com
- Online Ollama — catálogo de modelos abertos e documentação da API usada nesta disciplina. ollama.com
- Metodologia Oliveira, C.H.P. — "A metodologia de projeto aplicada ao curso superior de tecnologia". Educação Profissional: Ciência e Tecnologia , FIAP, 2007.
