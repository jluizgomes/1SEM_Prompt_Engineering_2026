## 📋 Roteiro do Lab

**Exercício da Aula 03**  
### Auditoria ética de um sistema de IA real ★ ★ ★

*Grupos de 3–4 · 15 minutos · Apresentação de 2 min por grupo*

1. Onde pode haver viés? Que grupos podem ser sistematicamente favorecidos ou prejudicados? Que dados de treinamento podem ter causado isso?
2. Quais dados são sensíveis? Que dados pessoais o sistema processa? Algum é dado sensível pela LGPD (saúde, raça, religião, biometria)?
3. O que violaria a LGPD? Identifique 1 cenário concreto em que o sistema poderia violar a LGPD como hoje é operado.
4. Proposta de guarda-chuva: Que 1 mudança técnica ou de processo reduziria os riscos identificados?

> **💡 Dica:**
>
> você não precisa saber como o sistema funciona por dentro — faça inferências pelo comportamento observável e pelas notícias sobre o sistema. O raciocínio crítico é mais importante do que precisão técnica aqui.
