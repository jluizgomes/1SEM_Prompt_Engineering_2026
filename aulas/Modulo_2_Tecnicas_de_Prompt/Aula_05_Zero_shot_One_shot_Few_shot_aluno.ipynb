```python
# Setup — não modifique
!pip install -q ollama
import ollama
from google.colab import userdata
```

```python
client = ollama.Client(
    host="https://api.ollama.com",
    headers={"Authorization": f"Bearer {userdata.get("OLLAMA_API_KEY")}"}
)

# ── TAREFA: classificador de URGÊNCIA de tickets de suporte ──
# Categorias: CRÍTICO, ALTO, MÉDIO, BAIXO

few_shot = f"""
Classifique a urgência do ticket: CRÍTICO, ALTO, MÉDIO ou BAIXO.

👉 LACUNA 1: escreva um exemplo de ticket CRÍTICO
Ticket: ___ → CRÍTICO

👉 LACUNA 2: escreva um exemplo de ticket ALTO
Ticket: ___ → ALTO

👉 LACUNA 3: escreva um exemplo de ticket MÉDIO
Ticket: ___ → MÉDIO

👉 LACUNA 4: escreva um exemplo de ticket BAIXO
Ticket: ___ → BAIXO

Ticket: "O sistema de pagamento está fora do ar para todos os clientes."
"""

# Chamada e resultado — não modifique
resp = client.chat(
    model="qwen3.6:27b",
    messages=[{"role": "user", "content": few_shot}]
)
print(f"Resultado: {resp["message"]["content"]}")
```
