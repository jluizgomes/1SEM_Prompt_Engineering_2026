```python
# Setup — não modifique
!pip install -q ollama
import ollama
from google.colab import userdata
```

```python
client = ollama.Client(
    host="https://api.ollama.com",
    headers={"Authorization": f"Bearer {userdata.get('OLLAMA_API_KEY')}"}
)

# TAREFA: decidir entre 2 opções de tecnologia com CoT few-shot
# Complete os passos de raciocínio dentro do prompt (não o código!)

cot_few_shot = """
Você é um arquiteto de software. Avalie qual tecnologia usar.
Mostre o raciocínio em passos antes de concluir.

--- Exemplo 1 ---
Situação: App de e-commerce com 10k usuários, dados de produtos e pedidos.
Raciocínio:
👉 Passo 1: descreva o tipo de dado (estruturado? não-estruturado?)

👉 Passo 2: identifique se há relacionamentos importantes

👉 Passo 3: avalie se escala é crítica agora ou futuramente

Conclusão: [banco relacional / NoSQL / outro]

--- Exemplo 2 ---
Situação: Sistema de logs de servidor, 1M eventos/dia, sem relacionamentos.
Raciocínio:
👉 Passo 1: tipo de dado deste cenário

👉 Passo 2: relacionamentos e consultas mais comuns

👉 Passo 3: escala e velocidade de escrita

Conclusão: [banco relacional / NoSQL / outro]

--- Situação a analisar ---
Situação: Rede social com perfis de usuários, posts, curtidas e seguidores.
Raciocínio:
"""

# Chamada — não modifique
resp = client.chat(
    model="qwen3.6:27b",
    messages=[{"role": "user", "content": cot_few_shot}]
)
print(resp["message"]["content"])
```
