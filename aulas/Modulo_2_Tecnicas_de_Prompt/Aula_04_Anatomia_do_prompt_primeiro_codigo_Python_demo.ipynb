### Slide 23 — Conceitos Python usados hoje — para referência

````python
      # VARIÁVEL — guarda um valor com um nome
modelo = "qwen3.6:27b"  # string (texto)
temperatura = 0.7         # float (número decimal)

# STRING — texto entre aspas. Triple quotes = múltiplas linhas
texto_curto = "uma linha"
texto_longo = """
primeira linha
segunda linha
"""

# DICIONÁRIO — pares chave:valor, acessados por colchetes
mensagem = {"role": "user", "content": "Olá"}
print(mensagem["role"])     # → "user"
print(mensagem["content"])  # → "Olá"

# LISTA — sequência de valores, acessada por índice [0]
mensagens = [mensagem]  # lista com 1 dicionário dentro

# CHAMADA DE FUNÇÃO/MÉTODO — nome() ou objeto.nome()
resposta = client.chat(model=modelo, messages=mensagens)

# F-STRING — string com variável dentro usando {}
print(f"Modelo usado: {modelo}")  # → "Modelo usado: qwen3.6:27b"

# PARÂMETRO NOMEADO (kwarg) — nome=valor dentro de uma chamada
resposta = client.chat(
    model=modelo, messages=mensagens,
    options={"num_predict": 150}  # max_tokens da Lacuna 3
)
````

### Slide 25 — 🚀 Demo — As 8 linhas explicadas uma a uma

````python
      # ── linha 1: instalar a biblioteca que faz a chamada à API ──
!pip install -q ollama

# ── linha 2: importar a biblioteca instalada ──
import ollama
from google.colab import userdata

# ── linha 3: pegar a chave de API dos Secrets do Colab ──
api_key = userdata.get("OLLAMA_API_KEY")

# ── linha 4: criar o cliente que vai se conectar à API ──
client = ollama.Client(
    host="https://api.ollama.com",
    headers={"Authorization": f"Bearer {api_key}"}
)

# ── linha 5: escrever o prompt com os 4 componentes ──
meu_prompt = """
Você é um professor universitário de tecnologia.     # contexto
Explique o que é inteligência artificial             # instrução
para um estudante de 1º ano que nunca programou.    # contexto
Use 1 analogia do cotidiano. Máximo de 80 palavras. # formato
"""

# ── linha 6–7: enviar o prompt e receber a resposta ──
resposta = client.chat(
    model="qwen3.6:27b",
    messages=[{"role": "user", "content": meu_prompt}]
)

# ── linha 8: exibir a resposta ──
print(resposta["message"]["content"])
````
