### Slide 06 — System prompt no código Python

```python
# Sem system prompt (como nas aulas anteriores)
resp = client.chat(
    model="qwen3.6:27b",
    messages=[{"role": "user", "content": "O que é machine learning?"}]
)

# COM system prompt — a persona é separada do pedido do usuário
resp = client.chat(
    model="qwen3.6:27b",
    messages=[
        {
            "role": "system",    # ← role "system"
            "content": """Você é um professor universitário de tecnologia.
Explique conceitos de forma clara e use analogias do cotidiano.
Nunca use jargão sem explicar antes. Público-alvo: alunos de 1º ano."""
        },
        {
            "role": "user",      # ← role "user" — o pedido do usuário
            "content": "O que é machine learning?"
        }
    ]
)
print(resp["message"]["content"])
```

### Slide 10 — Código novo — dicionário Python

### Slide 26 — 🚀 Demo Prática ao Vivo — Dicionário de personas + loop

```python
# Setup (igual às aulas anteriores — não repetido aqui)

# ── NOVO: dicionário Python ────────────────────────────────
# Dicionário: estrutura chave → valor, acessada por chave
personas = {
    "especialista": """Você é um pesquisador sênior em IA com doutorado pelo MIT.
Responda com precisão técnica. Use termos como transformers,
gradient descent e backpropagation quando relevante.
Não simplifique conceitos.""",

    "professor": """Você é um professor universitário de tecnologia
explicando para alunos de 1º ano. Use analogias simples do
cotidiano. Explique todo termo técnico antes de usar.
Tom encorajador e paciente.""",

    "jornalista": """Você é um jornalista investigativo especializado
em tecnologia. Escreva de forma objetiva e crítica. Questione
o hype. Mencione riscos e controvérsias. Formato: jornal de
qualidade, pirâmide invertida.""",
}

# Mesma pergunta para todas as personas
pergunta = "O que é inteligência artificial e por que está crescendo tanto?"

# ── Loop que testa cada persona ──────────────────────────
# personas.items() retorna pares (chave, valor)
for nome, system in personas.items():
    resp = client.chat(
        model="qwen3.6:27b",
        messages=[
            {"role": "system",  "content": system},
            {"role": "user",    "content": pergunta}
        ]
    )
    print(f"\n{'='*50}\nPERSONA: {nome.upper()}\n{'='*50}")
    print(resp["message"]["content"])
```
