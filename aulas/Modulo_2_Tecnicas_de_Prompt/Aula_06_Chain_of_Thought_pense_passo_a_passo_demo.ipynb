como CRÍTICO, ALTO ou NORMAL. Mostre seu raciocínio antes de concluir.

Ticket: "Sistema de pagamento fora do ar para todos os clientes."
Raciocínio:
- Impacto: todos os clientes afetados (escopo total)

Raciocínio:

- Funcionalidade: relatório = importante mas não core
- Urgência: pode aguardar até o próximo dia útil
Conclusão: NORMAL

Raciocínio:

### Slide 25 — 🚀 Demo Prática — Comparativo ao vivo com 3 variantes

````python
# Setup (igual às aulas anteriores)
!pip install -q ollama
import ollama
from google.colab import userdata
client = ollama.Client(
    host="https://api.ollama.com",
    headers={"Authorization": f"Bearer {userdata.get('OLLAMA_API_KEY')}"}
)

# Problema de lógica que o modelo frequentemente erra sem CoT
problema = """Um fazendeiro tem 17 ovelhas. Todas menos 9 morrem.
Quantas ovelhas o fazendeiro tem agora?"""

variantes = [
    ("SEM CoT",
     f"Responda diretamente: {problema}"),

    ("CoT ZERO-SHOT",
     f"Resolva o problema. Pense passo a passo antes de responder.\n\n{problema}"),

    ("CoT FEW-SHOT",
     """Problema: "Tenho 10 maçãs e como 3. Quantas restam?"
Raciocínio: Inicio com 10. Removo 3. 10 - 3 = 7.
Resposta: 7

Problema: "Havia 5 carros, 2 saíram. Quantos ficaram?"
Raciocínio: Inicio com 5. Removo 2. 5 - 2 = 3.
Resposta: 3

""" + f"Problema: {problema}\nRaciocínio:"),
]

for nome, prompt in variantes:
    resp = client.chat(
        model="qwen3.6:27b",
        messages=[{"role": "user", "content": prompt}]
    )
    print(f"\n{'='*40}\n{nome}:\n{resp['message']['content']}")
````

````python
# Setup — não modifique
!pip install -q ollama
import ollama
from google.colab import userdata
client = ollama.Client(
    host="https://api.ollama.com",
    headers={"Authorization": f"Bearer {userdata.get('OLLAMA_API_KEY')}"}
)

# TAREFA: decidir entre 2 opções de tecnologia com CoT few-shot
# Complete os passos de raciocínio dentro do prompt (não o código!)

cot_few_shot = """
Você é um arquiteto de software. Avalie qual tecnologia usar.
Mostre o raciocínio em passos antes de concluir.

--- Exemplo 1 ---
Situação: App de e-commerce com 10k usuários, dados de produtos e pedidos.
Raciocínio:
👉 Passo 1: descreva o tipo de dado (estruturado? não-estruturado?)

👉 Passo 2: identifique se há relacionamentos importantes

👉 Passo 3: avalie se escala é crítica agora ou futuramente

Conclusão: [banco relacional / NoSQL / outro]

--- Exemplo 2 ---
Situação: Sistema de logs de servidor, 1M eventos/dia, sem relacionamentos.
Raciocínio:
👉 Passo 1: tipo de dado deste cenário

👉 Passo 2: relacionamentos e consultas mais comuns

👉 Passo 3: escala e velocidade de escrita

Conclusão: [banco relacional / NoSQL / outro]

--- Situação a analisar ---
Situação: Rede social com perfis de usuários, posts, curtidas e seguidores.
Raciocínio:
"""

# Chamada — não modifique
resp = client.chat(
    model="qwen3.6:27b",
    messages=[{"role": "user", "content": cot_few_shot}]
)
print(resp["message"]["content"])
````
