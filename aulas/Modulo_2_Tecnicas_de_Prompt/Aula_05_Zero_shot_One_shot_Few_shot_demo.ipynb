### Slide 12 — Código novo desta aula — for loop e lista de tuplas

```python
# Lista = sequência entre [ ]
# Tupla = par fixo entre ( )
variantes = [
    ("ZERO-SHOT", zero_shot),  # tupla 1
    ("ONE-SHOT",  one_shot),   # tupla 2
    ("FEW-SHOT",  few_shot),   # tupla 3
]
# Cada tupla tem: (nome_da_variante, texto_do_prompt)
```

### Slide 12 — Código novo desta aula — for loop e lista de tuplas

```python
# O for desempacota cada tupla
for nome, prompt in variantes:
    # 1ª iteração: nome="ZERO-SHOT", prompt=zero_shot
    # 2ª iteração: nome="ONE-SHOT",  prompt=one_shot
    # 3ª iteração: nome="FEW-SHOT",  prompt=few_shot
    print(f"{nome}: ...")

# f-string: coloca o valor da variável dentro do texto
# f"ZERO-SHOT: {resultado}" → "ZERO-SHOT: NEUTRO"
```

### Slide 23 — Resumo — novos conceitos Python desta aula

```python
# LISTA — sequência mutável entre [ ]
frutas = ["maçã", "banana", "laranja"]
print(frutas[0])  # → "maçã"  (índice começa em 0)

# TUPLA — par fixo entre ( ). Não pode mudar depois de criada.
par = ("ZERO-SHOT", zero_shot)

# LISTA DE TUPLAS — útil para agrupar dados relacionados
variantes = [
    ("ZERO-SHOT", zero_shot),
    ("ONE-SHOT",  one_shot),
    ("FEW-SHOT",  few_shot),
]

# FOR LOOP — repete para cada item de uma sequência
for fruta in frutas:
    print(fruta)  # imprime "maçã", "banana", "laranja"

# FOR COM DESEMPACOTAMENTO — desempacota tupla automaticamente
for nome, prompt in variantes:
    print(f"Testando: {nome}")  # f-string insere o valor da variável

# F-STRING — coloca variáveis dentro do texto com { }
resultado = "POSITIVO"
print(f"Sentimento: {resultado}")  # → "Sentimento: POSITIVO"
```

### Slide 25 — 🚀 Demo Prática — Loop que testa 3 variantes ao vivo

```python
# Setup (igual à aula anterior)
!pip install -q ollama
import ollama
from google.colab import userdata
client = ollama.Client(
    host="https://api.ollama.com",
    headers={"Authorization": f"Bearer {userdata.get("OLLAMA_API_KEY")}"}
)

# ── 3 versões do prompt ──────────────────────────────────
zero_shot = """
Classifique o sentimento: POSITIVO, NEGATIVO ou NEUTRO.
Texto: "O pacote atrasou mas o produto está ótimo."
"""

one_shot = """
Classifique o sentimento: POSITIVO, NEGATIVO ou NEUTRO.
Texto: "Entrega rápida, adorei!" → POSITIVO
Texto: "O pacote atrasou mas o produto está ótimo."
"""

few_shot = """
Classifique o sentimento: POSITIVO, NEGATIVO ou NEUTRO.
Texto: "Entrega rápida, adorei!" → POSITIVO
Texto: "Produto quebrado, horrível." → NEGATIVO
Texto: "Chegou no prazo, nada especial." → NEUTRO
Texto: "O pacote atrasou mas o produto está ótimo."
"""

# ── testar as 3 variantes com um for loop ────────────────
variantes = [
    ("ZERO-SHOT", zero_shot),
    ("ONE-SHOT", one_shot),
    ("FEW-SHOT", few_shot),
]

for nome, prompt in variantes:
    resp = client.chat(
        model="qwen3.6:27b",
        messages=[{"role": "user", "content": prompt}]
    )
    print(f"{nome}: {resp["message"]["content"]}")
```
