## 📋 Roteiro do Lab

**Lab — Aula 12**  
### Adicionar guardrail ao CKP03 e testar os 3 ataques ★★

*Grupo CKP03 · 25 minutos · Google Colab*

1. Abra o notebook do CKP03 — o chatbot que o grupo entregou na Aula 11.
2. Adicione a função validar_input() numa nova célula antes de iniciar_chat(). Adapte a lista TERMOS_SUSPEITOS para incluir termos específicos do domínio do grupo.
3. Insira a verificação no loop — duas linhas antes do montar_prompt().
4. Teste os 3 ataques da demo — confirmar que o guardrail bloqueia cada um. Documente o resultado no notebook com um comentário.
5. Desafio — guardrail de output: após receber a resposta do modelo, verificar se ela contém alguma das palavras da lista TERMOS_SUSPEITOS. Se sim, substituir por uma mensagem padrão.

> **⚠️ Limitações do guardrail simples**
>
> Um atacante sofisticado pode contornar o filtro de palavras-chave com variações ortográficas, outro idioma, base64, etc. O guardrail com if/else elimina ataques básicos e oportunistas — não é solução completa para produção.

> **💡 Guardrail de output (desafio):**
>
> após resposta = resp["message"]["content"], adicionar: ok, _ = validar_input(resposta, max_chars=2000); if not ok: resposta = "Não posso exibir esta resposta."
