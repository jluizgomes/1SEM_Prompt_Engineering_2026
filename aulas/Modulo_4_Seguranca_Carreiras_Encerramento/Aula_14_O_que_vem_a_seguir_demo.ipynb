### Slide 08 — 🎬 Trailer 2 — RAG: chatbot que responde sobre documentos

````python
from langchain.document_loaders import PyPDFLoader
from langchain.text_splitter import RecursiveCharacterTextSplitter
from langchain_community.vectorstores import Chroma
from langchain_community.embeddings import OllamaEmbeddings

# 1. Carregar o documento PDF
docs = PyPDFLoader("manual_empresa.pdf").load()

# 2. Dividir em pedaços menores (chunks)
chunks = RecursiveCharacterTextSplitter(
    chunk_size=500, chunk_overlap=50
).split_documents(docs)

# 3. Gerar embeddings e salvar em vector store
db = Chroma.from_documents(
    chunks,
    OllamaEmbeddings(model="nomic-embed-text")
)

# 4. Buscar o trecho mais relevante e responder
pergunta = "Qual a política de férias?"
contexto = db.similarity_search(pergunta, k=3)
# → envia contexto + pergunta ao LLM → resposta baseada no documento
````
