## 📋 Roteiro do Lab

**Exercício — Aula 13**  
### Carta para o eu de 5 anos ★

*Individual · 15 minutos · Papel ou editor de texto*

1. Qual carreira você imagina ter em 5 anos? (pode ser vaga — "trabalhar com tecnologia", "criar um produto", "ter minha empresa")
2. Como IA estaria presente no seu trabalho daqui a 5 anos? O que você usaria de IA no dia a dia dessa carreira imaginada?
3. O que você construiu neste semestre que você não sabia que seria capaz de construir quando entrou?
4. O que você vai fazer diferente no próximo semestre para avançar mais rápido?
5. Uma habilidade que você quer dominar até o final do curso de CC.

> **📢 Compartilhamento voluntário**
>
> Ao final dos 15 min, quem quiser compartilha 1 das respostas com a turma. Sem obrigação — mas escutar o caminho dos outros costuma abrir perspectivas novas.

> **💡 Por que este exercício vale o tempo:**
>
> pesquisas em educação profissional mostram que estudantes com clareza de propósito têm desempenho e conclusão de curso significativamente maiores. 5 linhas agora podem economizar semestres de dúvida depois.
