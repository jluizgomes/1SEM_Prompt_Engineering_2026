## 📚 Referências da aula

- Prático LangChain Academy — Cursos gratuitos de LangChain e LangGraph. Começa pelo Quickstart e segue para RAG e Agentes. O próximo passo natural após este semestre. academy.langchain.com
- Prático Hugging Face Learn — Cursos gratuitos de NLP, LLMs e agentes com notebooks Colab. Excelente para aprofundar os fundamentos técnicos de transformers e embeddings. huggingface.co/learn
- Referência Anthropic Prompt Library + Engineering Guide — 100+ exemplos de prompts reais e o guia técnico de prompting mais completo disponível. docs.anthropic.com/pt/prompt-library
- Referência OWASP LLM Top 10 — Para manter a segurança em pauta conforme os projetos ficam mais complexos no 2º semestre. owasp.org/www-project-top-10-for-large-language-model-applications
- Livro Russell, S.; Norvig, P. — Inteligência Artificial: Uma Abordagem Moderna. 3ª ed. Pearson, 2016. A referência canônica — útil ao longo de todo o curso de CC, não só nesta disciplina.
- Comunidade Papers with Code — Para quando quiser ver como os avanços de pesquisa em IA se tornam código. Conecta papers a implementações reais. paperswithcode.com
- Livro Russell, S.; Norvig, P. — Inteligência Artificial: Uma Abordagem Moderna. 3ª ed. Pearson, 2016. Referência canônica da disciplina — usada nas Aulas 01, 09, 12, 13.
- Livro Muller, A. C.; Guido, S. — Introdução ao Aprendizado de Máquina com Python. Pearson, 2016. Fundamentos de Python para ML — referenciado no M3.
- Paper Brown, T. et al. — "Language Models are Few-Shot Learners." NeurIPS, 2020. Fundamento teórico das técnicas de few-shot e CoT — Aulas 05 e 06. arxiv.org/abs/2005.14165
- Online Anthropic — Prompt Library e Engineering Guide. Referência para boas práticas de prompting ao longo do M2. docs.anthropic.com/pt
- Online OWASP — Top 10 for Large Language Model Applications v2.0 (2025). Referência de segurança — Aula 12. owasp.org
- Livro Oliveira, C. H. P. — "A metodologia de projeto aplicada ao curso superior de tecnologia." Educação Profissional: Ciência e Tecnologia, v.2, n.1, 2007. Fundamento metodológico da avaliação por projetos desta disciplina.
