### Slide 25 — 🚀 Demo — Guardrail adicionado ao chatbot ao vivo

```
# ── GUARDRAIL DE INPUT ── adicionar ANTES do loop while True

# Palavras e frases associadas a tentativas de injection
TERMOS_SUSPEITOS = [
    "ignore", "esqueça", "esquece",
    "nova instrução", "novas instruções",
    "agora você é", "you are now",
    "sem restrições", "without restrictions",
    "repita suas instruções", "repeat your instructions",
    "system prompt",
]

def validar_input(texto, max_chars=500):
    """Verifica se o input do usuário parece seguro."""
    # Verificação 1: tamanho — inputs muito longos são suspeitos
    if len(texto) > max_chars:
        return False, "Mensagem muito longa. Por favor, seja mais conciso."

    # Verificação 2: termos de injection
    texto_lower = texto.lower()
    for termo in TERMOS_SUSPEITOS:
        if termo in texto_lower:
            return False, "Não consigo processar esse tipo de mensagem."

    # Input passou nas verificações
    return True, ""

# ── No loop do iniciar_chat(), antes do montar_prompt() ──
# Substituir a chamada direta por:
seguro, motivo = validar_input(entrada)
if not seguro:
    print(f"Assistente: {motivo}\n")
    continue                # pula para o próximo turno sem chamar o LLM

# Se passou, segue o fluxo normal
prompt_formatado = montar_prompt(entrada, especialidade)
```
