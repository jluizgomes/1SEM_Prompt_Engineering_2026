### Slide 09 — while True e break — o código novo desta aula

```python
# FOR — número fixo de iterações
for i in range(5):          # 5 vezes, sem surpresa
    print(i)

# WHILE TRUE — número desconhecido de iterações
while True:                 # roda para sempre...
    cmd = input("Digite: ")
    if cmd == "sair":
        break               # ...até o break
    print(f"Você digitou: {cmd}")

# .lower() — transforma em minúsculas
# "SAIR".lower() == "sair" → True
# Evita que "Sair" ou "SAIR" não funcionem
```

### Slide 19 — Resumo — Python novo desta aula

```python
# WHILE TRUE — loop infinito com condição de saída manual
while True:
    cmd = input("Digite algo (ou 'sair'): ")
    if cmd.lower() == "sair":
        break                # sai do while imediatamente
    print(f"Você disse: {cmd}")

# .lower() — converte string para minúsculas
print("SAIR".lower())       # → "sair"
print("Olá MUNDO".lower()) # → "olá mundo"

# O PADRÃO DO HISTÓRICO — acumular ambos os lados
historico = [{"role": "system", "content": "Você é assistente."}]

# A cada turno:
historico.append({"role": "user",      "content": entrada})   # antes
historico.append({"role": "assistant", "content": resposta})  # depois

# TRUNCAR SE FICAR LONGO (bônus — CKP03)
if len(historico) > 20:
    historico = historico[:1] + historico[-8:]
    # mantém system [0] + últimas 4 trocas (8 mensagens)
```

### Slide 25 — 🚀 Demo Prática ao Vivo — Loop de conversa com histórico (~20 linhas)

```python
# Setup + chamar_llm() da Aula 08 (copiar célula 1 e 2)
# [professor cola o setup e a função chamar_llm aqui]

# ── LOOP DE CONVERSA COM HISTÓRICO ──────────────────────
def iniciar_chat(system=""):
    """Loop de conversa com histórico acumulado."""
    historico = []

    # Adiciona system prompt se fornecido
    if system:
        historico.append({"role": "system", "content": system})

    print("Chat iniciado. Digite 'sair' para encerrar.\n")

    while True:                          # ← loop infinito

        # Passo 1: ler o input do usuário
        entrada = input("Você: ")

        # Passo 1b: condição de saída
        if entrada.lower() == "sair":
            print("Encerrando chat. Até logo!")
            break                          # ← encerra o while

        # Passo 2: adicionar ao histórico
        historico.append({"role": "user", "content": entrada})

        # Passo 3: chamar API com o histórico inteiro
        resp = _client.chat(
            model="qwen3.6:27b",
            messages=historico           # ← lista inteira, não só a última
        )

        # Passo 4: extrair o texto
        resposta = resp["message"]["content"]

        # Passo 5: adicionar resposta ao histórico
        historico.append({"role": "assistant", "content": resposta})

        # Exibir e repetir
        print(f"Assistente: {resposta}\n")
        print(f"[histórico: {len(historico)} mensagens]\n")

# Iniciar com persona de tutor de Python
iniciar_chat(system="Você é um tutor de Python. Seja breve e didático.")
```
