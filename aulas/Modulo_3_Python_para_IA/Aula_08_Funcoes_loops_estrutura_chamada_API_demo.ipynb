### Slide 12 — Novo conceito Python: enumerate()

````python
# Sem enumerate — você não sabe o índice
frutas = ["maçã", "banana", "laranja"]
for fruta in frutas:
    print(fruta)
# maçã / banana / laranja

# Com enumerate — você tem índice + valor
for i, fruta in enumerate(frutas, 1):
    print(f"{i}. {fruta}")
# 1. maçã / 2. banana / 3. laranja

# O "1" em enumerate(frutas, 1) é o início
# do contador — padrão é 0, usamos 1 para
# exibição mais natural
````

### Slide 21 — Resumo — Python novo desta aula

````python
# DEF — declarar função
def minha_funcao(param1, param2="padrão"):
    """Docstring: descreve o que a função faz."""
    resultado = param1 + " " + param2
    return resultado          # sem return → devolve None

# CHAMANDO — com e sem parâmetro opcional
minha_funcao("olá")          # → "olá padrão"
minha_funcao("olá", "mundo")  # → "olá mundo"

# IF PARA VERIFICAR STRING VAZIA
if param2:    # True se não vazio, False se ""
    print("tem conteúdo")

# LIST.APPEND — adicionar item ao final da lista
lista = []
lista.append("primeiro")    # → ["primeiro"]
lista.append("segundo")     # → ["primeiro", "segundo"]

# ENUMERATE — índice + valor no for loop
itens = ["a", "b", "c"]
for i, item in enumerate(itens, 1):  # começa em 1
    print(f"{i}. {item}")  # → 1. a / 2. b / 3. c
````

### Slide 25 — 🚀 Demo Prática ao Vivo — Construindo chamar_llm() do zero

````python
# ── SETUP (célula 1) ────────────────────────────────────
!pip install -q ollama
import ollama
from google.colab import userdata

_client = ollama.Client(
    host="https://api.ollama.com",
    headers={"Authorization": f"Bearer {userdata.get('OLLAMA_API_KEY')}"}
)

# ── A FUNÇÃO (célula 2) ──────────────────────────────────
def chamar_llm(prompt, system="", modelo="qwen3.6:27b"):
    """Envia um prompt ao LLM e retorna a resposta como texto."""
    mensagens = []

    # Só adiciona system prompt se não for vazio
    if system:
        mensagens.append({"role": "system", "content": system})

    # Sempre adiciona a mensagem do usuário
    mensagens.append({"role": "user", "content": prompt})

    # Chama o modelo e extrai só o texto da resposta
    resp = _client.chat(model=modelo, messages=mensagens)
    return resp["message"]["content"]

# ── TESTE SIMPLES (célula 3) ─────────────────────────────
resultado = chamar_llm("O que é uma função em Python? Responda em 1 linha.")
print(resultado)
````
