### Slide 08 — Além de variáveis — expressões dentro do {}

```python
# Variável simples
nome = "carlos"
print(f"Nome: {nome}")               # → "Nome: carlos"

# Método de string dentro do {}
print(f"Nome: {nome.title()}")        # → "Nome: Carlos"
print(f"Nome: {nome.upper()}")        # → "Nome: CARLOS"

# Operação matemática
limite = 100
print(f"Máximo: {limite * 2} palavras") # → "Máximo: 200 palavras"

# Condicional (expressão ternária)
formal = True
print(f"Tom: {'formal' if formal else 'informal'}")

# Aplicado a templates de prompt
eh_tecnico = False
publico    = "gestor"

prompt = f"""
Explique para um {publico}.
Nível técnico: {'alto' if eh_tecnico else 'baixo'}.
{'Inclua exemplos de código.' if eh_tecnico else 'Evite jargão técnico.'}
"""
```

### Slide 18 — Template + histórico — integração real

```python
# Template para chatbot especializado em Python
def montar_prompt(pergunta, nivel="iniciante"):
    return f"""
Pergunta sobre Python ({nivel}): {pergunta}
{'Explique como se eu tivesse 10 anos.' if nivel == 'iniciante' else ''}
Use 1 exemplo de código curto. Máximo 3 linhas de explicação.
"""

# No loop de conversa — monta o prompt antes de adicionar ao histórico
historico = [{"role": "system", "content": "Você é tutor de Python."}]

while True:
    entrada = input("Você: ")
    if entrada.lower() == "sair": break

    # Usa montar_prompt() para formatar antes de enviar
    prompt_formatado = montar_prompt(entrada, nivel="iniciante")

    # Adiciona ao histórico — o prompt formatado, não a entrada bruta
    historico.append({"role": "user", "content": prompt_formatado})

    resp = _client.chat(model="qwen3.6:27b", messages=historico)
    resposta = resp["message"]["content"]
    historico.append({"role": "assistant", "content": resposta})
    print(f"Assistente: {resposta}\n")
```

### Slide 25 — 🚀 Demo Prática ao Vivo — Gerador de e-mails com template (~15 linhas)

```python
# Setup + chamar_llm() da Aula 08 (copiar célula 1 e 2)

# ── A FUNÇÃO montar_prompt() ─────────────────────────────
def montar_prompt(
    destinatario,
    assunto,
    tom="formal",
    max_palavras=80,
    contexto=""
):
    """Monta um prompt para geração de e-mail com parâmetros variáveis."""
    return f"""
Você é um assistente de comunicação profissional.
Escreva um e-mail {tom} para {destinatario} sobre: {assunto}.
{'Contexto adicional: ' + contexto if contexto else ''}
Limite: {max_palavras} palavras.
Inclua: assunto, saudação, corpo e despedida.
Responda em português brasileiro.
"""

# ── TESTAR COM 3 CONTEXTOS ───────────────────────────────
cenarios = [
    {"destinatario": "Pedro",  "assunto": "atraso na entrega",  "tom": "formal"},
    {"destinatario": "Ana",    "assunto": "reunião de sexta",   "tom": "informal"},
    {"destinatario": "Suporte", "assunto": "produto com defeito", "tom": "reclamação"},
]

for c in cenarios:
    # Monta o prompt com os parâmetros do cenário
    prompt = montar_prompt(**c)
    # Envia usando chamar_llm() da Aula 08
    email = chamar_llm(prompt)
    print(f"\n──── {c['tom'].upper()} → {c['destinatario']} ────")
    print(email)
```
