### Slide 16 — Além do mínimo — melhorias que valem pontos extras no CKP03

````python
# Versão estendida do chatbot — extras opcionais
def iniciar_chat(especialidade, persona, max_turnos=20):
    historico = [{"role": "system", "content": persona}]
    turno = 0                     # extra 1: contador de turnos
    print(f"🤖 {especialidade} — máx {max_turnos} turnos\n")

    while True:
        if turno >= max_turnos:   # extra 2: limite automático
            print("Limite de turnos atingido.")
            break

        entrada = input("Você: ")
        if entrada.lower() == "sair": break

        prompt_formatado = montar_prompt(entrada, especialidade)
        historico.append({"role": "user", "content": prompt_formatado})

        # extra 3: janela deslizante — mantém system + últimos 10
        if len(historico) > 12:
            historico = historico[:1] + historico[-10:]

        resp     = _client.chat(model=MODELO, messages=historico)
        resposta = resp["message"]["content"]
        historico.append({"role": "assistant", "content": resposta})

        turno += 1
        print(f"[turno {turno} · {len(historico)} msgs] {resposta}\n")
````

### Slide 26 — 🚀 Demo Prática ao Vivo — Chatbot integrado (parte 1 de 2): setup e template

````python
# ── CÉLULA 2: Template de prompt (peça da Aula 10) ───────
def montar_prompt(pergunta, especialidade, estilo="direto e didático"):
    """Formata a pergunta do usuário com contexto da especialidade."""
    return f"""
Pergunta sobre {especialidade}: {pergunta}
Estilo de resposta: {estilo}.
Use exemplos concretos quando útil. Máximo 3 parágrafos.
"""

# Testar o template ANTES de integrar ao chatbot
teste = montar_prompt("o que é marinada?", "culinária brasileira")
print(teste)
# ← confirmar que o template ficou certo ANTES de chamar o LLM
````

### Slide 27 — 🚀 Demo — Chatbot integrado (parte 2 de 2): loop e integração

````python
# ── CÉLULA 3: Chatbot integrado — as 3 peças do M3 juntas
def iniciar_chat(especialidade, persona):
    """
    Chatbot especializado com histórico e template de prompt.
    Integra: montar_prompt (A10) + histórico messages (A09) + Ollama API (A08)
    """
    # Inicializa o histórico com a persona (system prompt) — A07
    historico = [{
        "role": "system",
        "content": persona
    }]

    print(f"\n🤖 Chatbot de {especialidade} iniciado.")
    print("   Digite 'sair' para encerrar.\n")

    # Loop de conversa — A09
    while True:

        # Passo 1: ler entrada do usuário
        entrada = input("Você: ")
        if entrada.lower() == "sair":
            print("Encerrando. Até logo!")
            break

        # Passo 2: usar template para formatar a pergunta — A10
        prompt_formatado = montar_prompt(entrada, especialidade)

        # Passo 3: adicionar ao histórico — A09
        historico.append({"role": "user", "content": prompt_formatado})

        # Passo 4: chamar a API com o histórico completo — A08
        resp = _client.chat(model=MODELO, messages=historico)
        resposta = resp["message"]["content"]

        # Passo 5: guardar a resposta no histórico — A09
        historico.append({"role": "assistant", "content": resposta})

        # Passo 6: exibir
        print(f"Assistente: {resposta}\n")

# ── Executar com domínio de culinária ────────────────────
iniciar_chat(
    especialidade="culinária brasileira",
    persona=(
        "Você é um chef especializado em culinária brasileira. "
        "Explique receitas com linguagem acessível, mencione variações "
        "regionais quando relevante e sugira substituições práticas."
    )
)
````
